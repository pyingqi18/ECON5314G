#### Example: $f_{Gaussian}(x) = \frac{1}{\sqrt{2\pi\sigma^2}}e^{-\frac{(x-\mu)^2}{2\sigma^2}}$
Here $\sigma$ must be positive. A production function should check that constraint and raise a clear error when it is violated.

```python
import numpy as np
import math

def my_fun4(x, mu, sig):
    y = 1/(np.sqrt(2*math.pi*sig**2))*np.exp(-(x-mu)**2/(2*sig**2))
    return y
```

```python
my_fun4(0, 0 ,1)
```

```python
import numpy as np
import math

def my_fun4_theta(x, theta):
    mu, sig = theta
    y = 1/(np.sqrt(2*math.pi*sig*sig))*np.exp(-(x-mu)**2/(2*sig**2))
    return y
```

```python
my_fun4_theta(0, (0, 1))
```

```python
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 200)
y = my_fun4(x, 1, 4)
fig, ax = plt.subplots()
ax.plot(x, y, 'r-', linewidth=2, label='my_fun4', alpha=0.6)
plt.show()
```

#### Applying a function repeatedly
A loop is useful when each step depends on the previous one. For independent array calculations, prefer NumPy vectorization when it is clear.

```python
for i in range(6):
    y = i**2
    print(i, y)
```

```python
x = [67,12,678,234,64,78,34,78,54]

y = np.zeros(len(x))
for i in range(len(x)):
    y[i] = x[i]**2
    print(x[i],y[i])
```

```python
def xsquare(x):
    y = np.zeros(len(x))
    for i in range(len(x)):
        y[i] = x[i]**2
    return y
```

```python
x = [6,2,8,3,4,7]
xsquare(x)
```

## 2. Optimization

Use [`scipy.optimize.minimize`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.minimize.html) with an objective and an initial guess. Always inspect `result.success`, `result.message`, and the objective value—not only `result.x`.

> **Watch for:** different starting values can lead to different local minima, and unconstrained optimization does not enforce parameter restrictions.

#### Example 1: minimize \begin{align} x^2 - 6x + 13 \end{align}

```python
from scipy.optimize import minimize
```

```python
def fun_1(x):

    return x**2 - 6*x + 13
```

```python
results = minimize(fun_1, -10)
```

```python
results
```

```python
from scipy.optimize import minimize

fun_1 = lambda x: x**2 - 6*x + 13
x0 = (10)

res = minimize(fun_1, x0, method = 'SLSQP')
print(res.x)
print(res.fun)
```

#### Example 2: minimize \begin{align} (x_1-1)^2 + (x_2 - 2.5)^2 \end{align}

```python
from scipy.optimize import minimize

fun_2 = lambda x: (x[0] - 1)**2 + (x[1] - 2.5)**2
x0 = (-50, -100)

res = minimize(fun_2, x0, method = 'SLSQP')
print(res.x)
print(res.fun)
```

#### Example 3: minimize \begin{align} x_1^2 - 2x_1x_2 + 8x_2^2 + (x_2 - 4)^2 \end{align}

```python
from scipy.optimize import minimize

fun_3 = lambda x: x[0]**2 - 2*x[0]*x[1] + 8*x[1]**2 + (x[1] - 4)**2

res = minimize(fun_3, (5, 10), method = 'SLSQP')
print(res.x)
print(res.fun)
```

```python
from scipy.optimize import minimize

def fun_3(x):
    z = x[0]**2 - 2*x[0]*x[1] + 8*x[1]**2 + (x[1] - 4)**2
    return z

res = minimize(fun_3, (6, 0), method = 'SLSQP')
print(res.x)
print(res.fun)
```

## 3. Integration

Use [`scipy.integrate.quad`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.integrate.quad.html) for one-dimensional integrals. It returns both the estimate and an estimate of numerical error.

#### Example : calculate \begin{align} \int_0^4 x^2 dx \end{align}

```python
from scipy import integrate
```

```python
f1 = lambda x: x**2
```

```python
y, err = integrate.quad(f1, 0, 4)
```

```python
y
```

```python
4**3/3
```

```python
err
```

#### Example : calculate \begin{align} \int_0^\infty e^{-x} dx \end{align}

```python
invexp = lambda x: np.exp(-x)
integrate.quad(invexp, 0, np.inf)
```
